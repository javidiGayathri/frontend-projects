import module from gemini package

```python
from google import genai
from google.colab import userdata
```

use client class to create an instance(for interacting with GeminiApi)


```python
client = genai.Client(api_key = userdata.get("GEMINI_API_KEY"))
print(client)
```

use client object for communication to GeminiLLM

```python
def study_assistant(user_prompt):
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=user_prompt
  )
  return response
```

```python
output = study_assistant("Explain LLM Architecture")
print(output.text)
```
